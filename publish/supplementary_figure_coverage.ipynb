```python
from matplotlib.gridspec import GridSpec


screenshots = [
    {"name": "top", "elevation": 90, "azimuth": 0, "distance": 300},
    {"name": "left", "elevation": 0, "azimuth": -90, "distance": 400},
    {"name": "left_medial", "elevation": 0, "azimuth": -90, "distance": 400},
    {"name": "front", "elevation": 0, "azimuth": 0, "distance": 300},
    {"name": "bottom", "elevation": -90, "azimuth": -180, "distance": 300},
    {"name": "right", "elevation": 0, "azimuth": 90, "distance": 400},
    {"name": "right_medial", "elevation": 0, "azimuth": 90, "distance": 400},
    {"name": "back", "elevation": 0, "azimuth": -180, "distance": 300},
]

friendly_names = {
    "top": "Top",
    "left": "Left",
    "left_medial": "Left inside-out",
    "front": "Front",
    "bottom": "Bottom",
    "right": "Right",
    "right_medial": "Right inside-out",
    "back": "Back",
}


fig = plt.figure(figsize=(12, 10), constrained_layout=True)
gs = GridSpec(2, 4, figure=fig)

for i, shot in enumerate(screenshots):
    ax = fig.add_subplot(gs[i // 4, i % 4])
    img = plt.imread(f"../output/brain/brain_view_{shot['name']}.png")
    ax.imshow(img)
    ax.axis("off")
    ax.set_title(friendly_names[shot["name"]])

plt.savefig("../output/figures/sfigure1.pdf", bbox_inches='tight', dpi=300)
plt.savefig("../output/figures/sfigure1.svg", bbox_inches='tight', dpi=300)
plt.show()
```
